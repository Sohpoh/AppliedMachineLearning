Perceptron takes binary inputs and produces one binary output. It works with linearly separable data. It uses a bias, and weighted inputs. It is very fast as the weight updates are simple. It's strength is with linearly separable data. The feature type it usually works best with is numerical since it works on weighted sums.


SVM's use hyperplanes to distinguish classes. It is slower than perceptron as it uses a quadratic optimization problem. It's strength is that it handles high dimensional data well. It is good at finding complex decision boundaries. The feature type is best with numerical but also good with nominal as well.

Decision trees use a tree like structure to make decisions, used for both classification and regression. The leaf nodes of this tree are the final outcomes. The speed is fast. It's strength is that it can handle both numerical and nominal features easily. The feature type as mentioned can be both numerical or nominal.

Random Forest is an algorithm which combines multiple decision trees to make predictions. It uses ensemble learning which basically means that it combines outputs from these decision trees. It is slower than a single than an individual decision tree, however these decision trees can be done in parallel. It's strength is that it is resilient to overfitting which improves accuracy. It can use both numerical and nominal features.

True Positive - correct prediction of a positive outcome
True Negetive - correct prediction of a negative outcome
False Positive - incorrect prediction of a positive outcome
False Negative - incorrect prediction of a negative outcome

Precision is how correctly a model can predict a positive outcome. It is given by the formula
(True Positives)/ (True Positives + False Positives)

Recall is the true positive rate, it is how well a model can find a positive instance
(True Positives) / (True Positives + False Negatives)

F1-score is the harmonic mean of Precision and Recall
(2 * Precision * Recall) / (Precision + Recall)

```python
import pandas as pd
import math
from itertools import product

df = pd.read_csv('Admission_Predict.csv').drop(columns=['Serial No.'])


def compute_stdev(data, mean):
    variance = sum((x- mean) **2 for x in data)/ len(data)
    return math.sqrt(variance)

def compute_mean(data):
    total =sum(data)
    count= len(data)
    return total/count

def compute_correlation(x, y):
    xMean, yMean =compute_mean(x), compute_mean(y)
    stdev_x, stdev_y =compute_stdev(x, xMean), compute_stdev(y, yMean)
    numerator = sum((x[i] - xMean)*(y[i]- yMean) for i in range(len(x)))
    denominator = len(x) * stdev_x*stdev_y
    return numerator / denominator if denominator != 0 else 0

correlation_matrix = {col: [] for col in df.columns}

for col1, col2 in product(df.columns, df.columns):
    correlation_value = compute_correlation(df[col1].tolist(), df[col2].tolist())
    correlation_matrix[col1].append(correlation_value)

correlation_df = pd.DataFrame(correlation_matrix)
correlation_df.index = df.columns

print("Correlation Matrix from Scratch")
print(correlation_df)
```
